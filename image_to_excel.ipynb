```python
from paddleocr import PaddleOCR
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
from huggingface_hub import login
from huggingface_hub import whoami
from transformers import pipeline
```

```python

#model A
ocr = PaddleOCR(
    text_detection_model_name="PP-OCRv5_server_det",
    text_recognition_model_name="PP-OCRv5_server_rec",
    use_doc_orientation_classify=True, # Disables document orientation classification model via this parameter
    use_doc_unwarping=True, # Disables text image rectification model via this parameter
    use_textline_orientation=True, # Disables text line orientation classification model via this parameter
)

# hugging face Login for accesing model B
login(token="hf_token")


#model B
model_id = "meta-llama/Llama-3.2-3B-Instruct"
pipe = pipeline(
    "text-generation",
    model=model_id,
    torch_dtype=torch.bfloat16,
    device_map="auto",
)
```

# prediction

## fields entered by the user


```python
# prediction
#1. user upload image and give field to extract
from google.colab import files
uploaded=files.upload()
fields = [
    "invoice_number",
    "gstin",
    "amount"
]
```

## backend

```python
import ast
import os
from openpyxl import Workbook, load_workbook
```

```python
file_name = list(uploaded.keys())[0]
print(file_name)
file_path = "/content/" + file_name
print(file_path)

result = ocr.predict(file_path)
save_path="output/extraProcessing/"

   ##########################   MODEL A ############################
#detecting image
for res in result:
    res.print()
    res.save_to_img(save_path)
    res.save_to_json(save_path)

#text ectracted
text=[]
for res in result:
    print(res.keys())
    print(res["rec_texts"])
    text.append(res["rec_texts"])

#converted in raw text
raw_text=" ".join(text[0])
raw_text



   ##########################   MODEL B  ############################
#prompt for model B
ocr_text = raw_text
user_fields=fields

system_prompt="""You are an invoice information extraction assistant.

The OCR text may contain spelling mistakes or OCR errors.
Correct obvious OCR mistakes using context.

IMPORTANT:
- Do not modify GSTINs.
- Do not modify invoice numbers.
- Do not modify dates.
- Do not modify amounts.
- return as dictionary
-  dont give anyother output except the field"""



user_prompt = f"""

Fields to extract:
{', '.join(user_fields)}

OCR Text:
{ocr_text}
"""

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
outputs = pipe(
    messages,
    max_new_tokens=256,
)

data=outputs[0]["generated_text"][-1]["content"]

data_dict=ast.literal_eval(data)

                      ########### excel file #######

file_path="output.xlsx"
if os.path.exists(file_path):
  #loadind exsiting workbook
  wb=load_workbook(file_path)
  ws=wb.active

else:
  wb=Workbook()
  ws=wb.active
  ws.append(user_fields)#header



row=ws.max_row+1
for col,field in enumerate(user_fields,start=1):
  ws.cell(row=row,column=col,value=data_dict[field])

wb.save(file_path)
print("file saved")
```
