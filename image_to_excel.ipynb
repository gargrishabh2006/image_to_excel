```python
from paddleocr import PaddleOCR
import torch
from huggingface_hub import login
from huggingface_hub import whoami
from transformers import pipeline
from google.colab import files
import ast
import os
from openpyxl import Workbook, load_workbook
```

# model importing



```python

#model A
ocr = PaddleOCR(
    text_detection_model_name="PP-OCRv5_server_det",
    text_recognition_model_name="PP-OCRv5_server_rec",
    use_doc_orientation_classify=True, # Disables document orientation classification model via this parameter
    use_doc_unwarping=True, # Disables text image rectification model via this parameter
    use_textline_orientation=True, # Disables text line orientation classification model via this parameter
)

# hugging face Login for accesing model B
login(token="your hf_token")


#model B
model_id = "meta-llama/Llama-3.2-3B-Instruct"
pipe = pipeline(
    "text-generation",
    model=model_id,
    torch_dtype=torch.bfloat16,
    device_map="auto",
)
```

# prediction

## fields entered by the user


```python
# prediction
#1. user upload image and give field to extract

uploaded=files.upload()
fields = ["Qty","Total","Bill To"]
```

## forward propagation


```python
file_name = list(uploaded.keys())[0]
print(file_name)
file_path = "/content/" + file_name
print(file_path)

result = ocr.predict(file_path)
save_path="output/extraProcessing/"

   ##########################   MODEL A ############################
#detecting image
for res in result:
    res.print()
    res.save_to_img(save_path)
    res.save_to_json(save_path)

#text ectracted

ocr_items = []

for res in result:
    texts = res["rec_texts"]
    boxes = res["rec_boxes"]

    for text, box in zip(texts, boxes):

        x1, y1, x2, y2 = map(int, box)

        center_x = (x1 + x2) / 2
        center_y = (y1 + y2) / 2

        ocr_items.append({
            "text": text,
            "x": center_x,
            "y": center_y,
            "box": [x1, y1, x2, y2]
        })

# Sort roughly top-to-bottom, then left-to-right
ocr_items.sort(key=lambda x: (x["y"], x["x"]))

ocr_text_with_position = ""

for item in ocr_items:
    ocr_text_with_position += (
        f"TEXT: {item['text']} | "
        f"X: {item['x']:.0f} | "
        f"Y: {item['y']:.0f}\n"
    )



   ##########################   MODEL B  ############################
#prompt for model B
ocr_text = ocr_text_with_position
user_fields=fields

system_prompt = """
You are an invoice information extraction assistant.

You will receive OCR text along with the X and Y position of each
detected text element.

The X coordinate represents the horizontal position on the invoice.
The Y coordinate represents the vertical position.

IMPORTANT RULES:

1. Use the spatial position of OCR elements to determine which
   column and row they belong to.

2. Do NOT assume that numbers appearing first in OCR text are Qty.

3. For table data, identify the column headers such as:
   Particulars, Qty, Rate, Amount, Tax, etc.

4. A number below the Qty column must be treated as Qty.
   A number below the Rate column must be treated as Rate.
   A number below the Amount column must be treated as Amount.

5. Preserve numeric values exactly as detected whenever possible.

6. Do not swap Qty, Rate, or Amount based only on their numerical values.

7. Correct obvious OCR spelling errors only when the correction is
   clear from context.

8. Do not modify GSTINs.
9. Do not modify invoice numbers.
10. Do not modify dates.
11. Do not modify amounts unless the OCR clearly contains an error.

12. Return output as dictionary with correct opening and closing brackets and key value format.
13. Do not provide explanations or any other text.
14. each field will have only 1 value (numeric)
"""
user_prompt = f"""
Fields to extract:
{', '.join(user_fields)}

OCR DATA WITH POSITIONS:
{ocr_text}
"""

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
outputs = pipe(
    messages,
    max_new_tokens=256,
)

data=outputs[0]["generated_text"][-1]["content"]

data_dict=ast.literal_eval(data)

                      ########### excel file #######

file_path="output.xlsx"
if os.path.exists(file_path):
  #loadind exsiting workbook
  wb=load_workbook(file_path)
  ws=wb.active

else:
  wb=Workbook()
  ws=wb.active
  ws.append(user_fields)#header



row=ws.max_row+1
for col,field in enumerate(user_fields,start=1):
  ws.cell(row=row,column=col,value=data_dict[field])

wb.save(file_path)
print("file saved")
```
